```python
from math import pi
```

```python
x=np.linspace(0,2*np.pi,50)
x
```

```python
from numpy import *
```

```python
y_sin=np.sin(x)
y_sin
```

```python
y_cos=np.cos(x)
y_cos
```

```python
plt.figure(figsize=(10,6))
plt.grid(True,linestyle='--',alpha=0.7)
```

```python
plt.plot(x,y_sin,color='#FF6B6B',linewidth=2.5,label='sin(x)')
plt.plot(x,y_cos,color='#4ECDC4',linewidth=2.5,label='cos(x)')
```

```python
plt.axhline(y=0,color='black',linewidth=2.5,alpha=0.8)
plt.axvline(x=np.pi,color='gray',linestyle=':',alpha=0.6,label='x=Π')
plt.axhline(y=1,color='gray',linestyle=':',alpha=0.6)
plt.axhline(y=-1,color='gray',linestyle=':',alpha=0.6)
```

```python
#画布
plt.figure(figsize=(10,6))
plt.grid(True,linestyle='--',alpha=0.7)
#
plt.plot(x,y_sin,color='#FF6B6B',linewidth=2.5,label='sin(x)')
plt.plot(x,y_cos,color='#4ECDC4',linewidth=2.5,label='cos(x)')
#辅助线
plt.axhline(y=0,color='black',linewidth=2.5,alpha=0.8)
plt.axvline(x=np.pi,color='gray',linestyle=':',alpha=0.6,label='x=Π')
plt.axhline(y=1,color='gray',linestyle=':',alpha=0.6)
plt.axhline(y=-1,color='gray',linestyle=':',alpha=0.6)
#图例
plt.legend(fontsize=11,loc='upper right')
plt.title('sin cos',fontsize=14,pad=15)
plt.xlabel('X',fontsize=12)
plt.ylabel('Y',fontsize=12)

plt.xticks([0,np.pi/2,np.pi,3*np.pi/2,2*np.pi],['0','Π/2','Π','3Π/2','2Π'],fontsize=10)
#自动调整
plt.tight_layout()
plt.show()
```
