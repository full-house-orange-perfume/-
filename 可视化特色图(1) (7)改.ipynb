```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

# 城市名称和人口数据
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Philadelphia', 'San Antonio', 'San Diego', 'Dallas', 'San Jose']
population = [8.4, 3.9, 2.7, 2.3, 1.7, 1.6, 1.5, 1.4, 1.3, 1.0]

# 优化颜色（用统一色系渐变，避免杂乱）
colors = plt.cm.Set3(np.linspace(0, 1, len(cities)))
# 简化图案（保留5种常用图案循环，避免视觉混乱）
patterns = ['/', '\\', '+', 'x', 'o'] * 2  # 循环使用

# 创建横向条形图（barh：horizontal bar）
fig, ax = plt.subplots(figsize=(11, 8), dpi=150)  # 1. 加宽图的宽度（从10→11），给文字留空间

# 绘制条形图时，调整条形位置，给Y轴文字留更多左侧空间
bars = ax.barh(cities, population, color=colors, hatch=patterns, edgecolor='black', linewidth=0.5, 
               left=0.3)  # 2. left=0.3：让条形整体右移，左侧留出空白放倾斜文字

# 添加数据标签（放在条形右侧，对齐更美观）
for bar, pop in zip(bars, population):
    plt.text(bar.get_width() + 0.1, bar.get_y() + bar.get_height()/2, 
             f'{pop}M', va='center', fontsize=9, fontweight='bold')

# 优化标题和标签样式
ax.set_title('Population of Major U.S. Cities (2023)', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Population (Millions)', fontsize=12, fontweight='500', labelpad=10)
ax.set_ylabel('City', fontsize=12, fontweight='500', labelpad=15)  # 3. 增加Y轴标签与刻度的间距

# -------------------------- 关键优化：纵坐标文字倾斜+避免遮挡 --------------------------
# rotation=30：倾斜30°；ha='left'：左对齐；va='center'：垂直居中
# fontsize=10：合适字号，避免过大
ax.set_yticklabels(cities, rotation=30, ha='left', va='center', fontsize=10)
# 4. 调整Y轴刻度的左侧间距（默认是0.1，调大到0.05更紧凑，或0.15更宽松）
ax.tick_params(axis='y', pad=10)  # pad=10：增加刻度文字与轴的距离

# 优化图例（简化为单个示例，避免多色块冗余）
legend_element = Patch(facecolor=colors[0], hatch=patterns[0], edgecolor='black', label='Population')
ax.legend(handles=[legend_element], loc='lower right', fontsize=10)

# 设置X轴范围，给数据标签留空间
ax.set_xlim(0, max(population) + 1)

# 优化网格线（只保留垂直网格，更整洁）
ax.grid(axis='x', linestyle='--', alpha=0.5, color='gray')

# 5. 手动调整图表边距，确保所有元素都能显示
plt.subplots_adjust(left=0.15, right=0.95, top=0.92, bottom=0.08)  # left=0.15：加大左侧边距

# 保存图片（指定格式为png，名称更清晰）
plt.savefig("./1横向城市人口条形图_无遮挡版.png", dpi=600, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# -------------------------- 1. 数据准备 --------------------------
categories = ['Category A', 'Category B', 'Category C']
data1 = [10, 15, 12]
data2 = [8, 11, 14]
data3 = [12, 9, 11]
data4 = [6, 14, 10]
data_list = [data1, data2, data3, data4]
labels = ['Data 1', 'Data 2', 'Data 3', 'Data 4']

# -------------------------- 2. 全局样式配置（提前设置，避免冲突） --------------------------
# 清爽网格样式，弱化网格不抢焦点
sns.set_style('whitegrid', {'grid.linestyle': '--', 'grid.alpha': 0.3, 'grid.color': '#cccccc'})
# 解决字体问题，提升文字质感
plt.rcParams['font.sans-serif'] = ['Arial', 'SimHei']  # 英文/中文兼容
plt.rcParams['axes.unicode_minus'] = False

# -------------------------- 3. 图表基础设置 --------------------------
bar_width = 0.18  # 适度加宽柱宽，平衡清晰与拥挤
x = np.arange(len(categories))  # 分类基准位置
# 调整画布尺寸，适配多组条形
fig, ax = plt.subplots(figsize=(11, 7), dpi=150)

# -------------------------- 4. 优化配色与纹理（清晰区分+视觉舒适） --------------------------
# 专业柔和色系，区分度高不刺眼
colors = ['#4A90E2', '#F5A623', '#27AE60', '#E74C3C']
# 简洁纹理，避免杂乱（纹理与颜色搭配增强区分）
hatches = ['//', '\\\\', '--', '..']

# -------------------------- 5. 绘制多组条形图（循环优化，避免重复代码） --------------------------
for i, (data, label, color, hatch) in enumerate(zip(data_list, labels, colors, hatches)):
    # 计算每组条形位置，确保居中对齐
    x_pos = x + (i - 1.5) * bar_width
    # 绘制条形，细节优化增强质感
    bars = ax.bar(
        x_pos, data, width=bar_width,
        label=label, color=color, alpha=0.85,  # 透明度适中，避免厚重
        edgecolor='white', linewidth=1.2,     # 白色边框，增强立体层次感
        hatch=hatch                           # 纹理简化，提升区分度
    )

    # -------------------------- 6. 数据标签优化（清晰不遮挡） --------------------------
    for bar, value in zip(bars, data):
        height = bar.get_height()
        # 标签位置紧贴条形顶部，加粗清晰
        ax.text(
            bar.get_x() + bar.get_width()/2, height + 0.2,
            str(value), ha='center', va='bottom',
            fontsize=10, fontweight='bold', color='#333333'
        )

# -------------------------- 7. 标题、坐标轴优化（层级清晰） --------------------------
# 标题：加粗放大，与内容拉开间距
ax.set_title('Styled Multi-Bar Chart', fontsize=16, fontweight='bold', pad=25, color='#2C3E50')
# 坐标轴标签：加大字号，优化颜色，增加间距
ax.set_xlabel('Categories', fontsize=12, fontweight='500', color='#34495E', labelpad=12)
ax.set_ylabel('Values', fontsize=12, fontweight='500', color='#34495E', labelpad=12)
# X轴刻度标签：加大字号，提升可读性
ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=11, color='#34495E')
# Y轴范围：留足顶部空间，避免标签溢出
ax.set_ylim(0, max(max(data) for data in data_list) + 1.5)

# -------------------------- 8. 图例优化（清晰不遮挡） --------------------------
ax.legend(
    loc='upper right', frameon=True, fancybox=True, shadow=True,
    framealpha=0.95, fontsize=10, labelspacing=0.6,
    bbox_to_anchor=(0.98, 0.98)  # 精准定位右上角，不遮挡图表
)

# -------------------------- 9. 细节打磨（提升精致感） --------------------------
# 隐藏顶部和右侧边框，清爽风格
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
# 保留的边框：变细改色，弱化存在感
ax.spines['left'].set_linewidth(0.8)
ax.spines['left'].set_color('#BDC3C7')
ax.spines['bottom'].set_linewidth(0.8)
ax.spines['bottom'].set_color('#BDC3C7')
# Y轴刻度：优化颜色和字号
ax.tick_params(axis='y', colors='#7F8C8D', labelsize=10)

# -------------------------- 10. 布局调整与保存（高清无截断） --------------------------
plt.tight_layout()  # 自动调整布局
# 保存高清图片，保留白色背景
plt.savefig(
    "./多条形图_优化清晰版.png", dpi=600, bbox_inches='tight',
    facecolor='white', edgecolor='none'
)
plt.show()
```

2.折线图

```python
import matplotlib.pyplot as plt
import numpy as np

# -------------------------- 1. 数据准备 --------------------------
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(2 * x)
y4 = np.cos(2 * x)

# -------------------------- 2. 全局样式配置 --------------------------
plt.rcParams['font.sans-serif'] = ['Arial', 'SimHei']  # 支持中英文
plt.rcParams['axes.unicode_minus'] = False  # 正常显示负号
plt.rcParams['axes.linewidth'] = 0.8  # 坐标轴线条宽度

# -------------------------- 3. 图表基础设置（预留右侧空间放图例） --------------------------
# 用gridspec调整子图布局，给右侧图例留1.5的宽度
fig = plt.figure(figsize=(13, 7), dpi=150)
gs = fig.add_gridspec(1, 2, width_ratios=[10, 1.5], hspace=0)
ax = fig.add_subplot(gs[0, 0])  # 主图表占左侧10份空间

# -------------------------- 4. 优化配色与线型 --------------------------
colors = ['#4A90E2', '#50C878', '#F5A623', '#D62728']
linestyles = ['-', '--', '-.', ':']
linewidths = [2.5, 2.2, 2.2, 2.2]
labels = ['Sin(x)', 'Cos(x)', 'Sin(2x)', 'Cos(2x)']

# 绘制折线
for i, (y, color, ls, lw, label) in enumerate(zip([y1, y2, y3, y4], colors, linestyles, linewidths, labels)):
    ax.plot(x, y, label=label, color=color, linestyle=ls, linewidth=lw, alpha=0.9)

# -------------------------- 5. 优化阴影区域 --------------------------
ax.fill_between(x, y1, y2, where=(y1 >= y2), color=colors[0], alpha=0.15, label='Sin(x) ≥ Cos(x)')
ax.fill_between(x, y1, y2, where=(y1 < y2), color=colors[1], alpha=0.15, label='Sin(x) < Cos(x)')

# -------------------------- 6. 标题与坐标轴优化 --------------------------
ax.set_title('Comparison of Trigonometric Functions', fontsize=18, fontweight='bold', pad=25, color='#2C3E50')
ax.set_xlabel('X-axis (Radians)', fontsize=13, fontweight='500', color='#34495E', labelpad=12)
ax.set_ylabel('Y-axis', fontsize=13, fontweight='500', color='#34495E', labelpad=12)

ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π'], fontsize=11, color='#34495E')
ax.set_ylim(-1.3, 1.3)
ax.set_yticks(np.arange(-1, 1.1, 0.5))
ax.set_yticklabels(np.arange(-1, 1.1, 0.5), fontsize=10, color='#34495E')

# -------------------------- 7. 参考线与注释优化 --------------------------
ax.axhline(y=0, color='#BDC3C7', linestyle=':', linewidth=1.2)
ax.axvline(x=np.pi, color='#BDC3C7', linestyle=':', linewidth=1.2)

# 注释框不遮挡线条
ax.text(5.2, 0.7, 'Sin(x) ≥ Cos(x)', fontsize=11, color=colors[0], fontweight='500',
        bbox=dict(boxstyle='round,pad=0.4', facecolor='white', alpha=0.8, edgecolor=colors[0], linewidth=1.5))
ax.text(1.0, -0.9, 'Sin(x) < Cos(x)', fontsize=11, color=colors[1], fontweight='500',
        bbox=dict(boxstyle='round,pad=0.4', facecolor='white', alpha=0.8, edgecolor=colors[1], linewidth=1.5))

# -------------------------- 8. 图例放在右侧外部（关键调整） --------------------------
# 获取所有图例元素
handles, labels_leg = ax.get_legend_handles_labels()
# 在右侧子图位置添加图例
fig.legend(handles, labels_leg, loc='center right', bbox_to_anchor=(0.97, 0.5),
           fontsize=10.5, labelspacing=0.8, frameon=True, fancybox=True, shadow=True, framealpha=0.95)

# -------------------------- 9. 网格与边框优化 --------------------------
ax.grid(True, linestyle='--', alpha=0.3, color='#BDC3C7')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#BDC3C7')
ax.spines['bottom'].set_color('#BDC3C7')

# -------------------------- 10. 布局调整与保存 --------------------------
plt.tight_layout()
# 保存时调整bbox_inches，确保图例完整显示
plt.savefig(
    "./折线图_图例不遮挡版.png", dpi=600, bbox_inches='tight',
    facecolor='white', edgecolor='none'
)
plt.show()
```

6.箱线图

```python
# 生成示例数据
np.random.seed(42)
data = [np.random.normal(0, 1, 200) for _ in range(6)]

# 创建箱线图
plt.figure(figsize=(12, 8))
bp = plt.boxplot(data, labels=['Group 1', 'Group 2', 'Group 3', 'Group 4', 'Group 5', 'Group 6'], patch_artist=True)

# 自定义箱体颜色
colors = ['lightblue', 'lightgreen', 'lightcoral', 'lightyellow', 'lightpink', 'lightsalmon']
for box, color in zip(bp['boxes'], colors):
    box.set(facecolor=color)

# 自定义离群值样式
for flier in bp['fliers']:
    flier.set(marker='o', markerfacecolor='red', markersize=8, linestyle='none')

# 自定义中位数线样式
for median in bp['medians']:
    median.set(linestyle='-', linewidth=2, color='green')

# 添加标签和标题
plt.xlabel('Groups')
plt.ylabel('Values')
plt.title('Complex and Beautiful Box Plot with Multiple Groups')

# 设置y轴限制
plt.ylim(-3, 3)

# 显示图形
plt.savefig("./箱线图", dpi=600)
plt.show()
```
